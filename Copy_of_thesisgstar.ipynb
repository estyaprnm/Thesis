#menyamakan total hari

```r
samakan_hari <- function(data){

  data$acq_date <- as.Date(data$acq_date)

  data_harian <- data %>%
    group_by(acq_date) %>%
    summarise(
      jumlah_hotspot = n()
    )

  hasil <- tanggal_lengkap %>%
    left_join(data_harian, by = "acq_date") %>%
    replace_na(list(jumlah_hotspot = 0))

  return(hasil)
}

kalbar_fix  <- samakan_hari(kalbar)
kalteng_fix <- samakan_hari(kalteng)
kalsel_fix  <- samakan_hari(kalsel)
kaltim_fix  <- samakan_hari(kaltim)
kaltara_fix <- samakan_hari(kaltara)
```

#uji heterogenitas

```r
# =========================================
# INDEKS GINI SESUAI RUMUS REFERENSI
# =========================================

gini_rumus <- function(K){

  # Hilangkan NA jika ada
  K <- K[!is.na(K)]

  # Urutkan dari terbesar ke terkecil
  K <- sort(K, decreasing = TRUE)

  # Banyak pengamatan
  n <- length(K)

  # Rata-rata
  K_bar <- mean(K)

  # Indeks urutan
  i <- 1:n

  # Menghitung Indeks Gini
  IG <- 1 + (1/n) -
    (2 / (n^2 * K_bar)) * sum(i * K)

  return(IG)
}
gini_K1 <- gini_rumus(kalimantanfixx$K1)
gini_K2 <- gini_rumus(kalimantanfixx$K2)
gini_K3 <- gini_rumus(kalimantanfixx$K3)
gini_K4 <- gini_rumus(kalimantanfixx$K4)
gini_K5 <- gini_rumus(kalimantanfixx$K5)

gini_K1
gini_K2
gini_K3
gini_K4
gini_K5
```

```r
library(forecast)

# ==============================
# NILAI AWAL + 1 AGAR POSITIF
# ==============================

Z1 <- kalimantanfixx$K1 + 1
Z2 <- kalimantanfixx$K2 + 1
Z3 <- kalimantanfixx$K3 + 1
Z4 <- kalimantanfixx$K4 + 1
Z5 <- kalimantanfixx$K5 + 1


# ==============================
# LAMBDA SEBELUM TRANSFORMASI
# ==============================

lambda1 <- BoxCox.lambda(Z1)
lambda2 <- BoxCox.lambda(Z2)
lambda3 <- BoxCox.lambda(Z3)
lambda4 <- BoxCox.lambda(Z4)
lambda5 <- BoxCox.lambda(Z5)

lambda1
lambda2
lambda3
lambda4
lambda5


# ==============================
# TRANSFORMASI BOX-COX
# menggunakan lambda masing-masing
# ==============================

Z1_trans <- Z1^lambda1
Z2_trans <- Z2^lambda2
Z3_trans <- Z3^lambda3
Z4_trans <- Z4^lambda4
Z5_trans <- Z5^lambda5


# ==============================
# CEK LAMBDA SETELAH TRANSFORMASI
# ==============================

lambda1_baru <- BoxCox.lambda(Z1_trans)
lambda2_baru <- BoxCox.lambda(Z2_trans)
lambda3_baru <- BoxCox.lambda(Z3_trans)
lambda4_baru <- BoxCox.lambda(Z4_trans)
lambda5_baru <- BoxCox.lambda(Z5_trans)

lambda1_baru
lambda2_baru
lambda3_baru
lambda4_baru
lambda5_baru
```
