#menyamakan total hari

```r
samakan_hari <- function(data){

  data$acq_date <- as.Date(data$acq_date)

  data_harian <- data %>%
    group_by(acq_date) %>%
    summarise(
      jumlah_hotspot = n()
    )

  hasil <- tanggal_lengkap %>%
    left_join(data_harian, by = "acq_date") %>%
    replace_na(list(jumlah_hotspot = 0))

  return(hasil)
}

kalbar_fix  <- samakan_hari(kalbar)
kalteng_fix <- samakan_hari(kalteng)
kalsel_fix  <- samakan_hari(kalsel)
kaltim_fix  <- samakan_hari(kaltim)
kaltara_fix <- samakan_hari(kaltara)
```

```r
# ================================
# VISUALISASI KORELASI DI PETA KALIMANTAN
# ================================

# Matriks korelasi
corr_mat <- matrix(c(
  1.0000000, 0.5835895, 0.4003273, 0.4617910, 0.3752124,
  0.5835895, 1.0000000, 0.9105719, 0.7169957, 0.3199425,
  0.4003273, 0.9105719, 1.0000000, 0.7938442, 0.3107967,
  0.4617910, 0.7169957, 0.7938442, 1.0000000, 0.6577895,
  0.3752124, 0.3199425, 0.3107967, 0.6577895, 1.0000000
), nrow = 5, byrow = TRUE)

colnames(corr_mat) <- rownames(corr_mat) <- c("K1","K2","K3","K4","K5")

# Posisi titik provinsi (disusun agar mirip peta Kalimantan)
lokasi <- data.frame(
  wilayah = c("K1", "K2", "K3", "K4", "K5"),
  nama = c("Kalbar", "Kalteng", "Kalsel", "Kaltim", "Kaltara"),
  x = c(1.2, 2.6, 2.7, 4.3, 4.4),
  y = c(2.2, 1.9, 1.0, 2.1, 3.4)
)

# Plot kosong
plot(NA,
     xlim = c(0.5, 5),
     ylim = c(0.5, 4),
     xlab = "",
     ylab = "",
     axes = FALSE,
     main = "Visualisasi Korelasi Antar Wilayah Kalimantan")

# Tambahkan garis hubungan korelasi
for(i in 1:4){
  for(j in (i+1):5){

    x1 <- lokasi$x[i]
    y1 <- lokasi$y[i]
    x2 <- lokasi$x[j]
    y2 <- lokasi$y[j]

    r <- corr_mat[i, j]

    # warna garis berdasarkan kekuatan korelasi
    warna <- if(r >= 0.8){
      "red"
    } else if(r >= 0.6){
      "orange"
    } else if(r >= 0.4){
      "gold"
    } else {
      "gray70"
    }

    # gambar garis
    segments(x1, y1, x2, y2,
             col = warna,
             lwd = 1 + 5*r)

    # tulis nilai korelasi di tengah garis
    text((x1+x2)/2, (y1+y2)/2,
         labels = round(r, 3),
         cex = 0.8,
         pos = 3)
  }
}

# Tambahkan titik wilayah
points(lokasi$x, lokasi$y,
       pch = 19,
       cex = 2,
       col = "blue")

# Tambahkan label K1-K5
text(lokasi$x, lokasi$y,
     labels = lokasi$wilayah,
     pos = 1,
     cex = 1.1,
     font = 2)

# Tambahkan nama provinsi
text(lokasi$x, lokasi$y + 0.18,
     labels = lokasi$nama,
     cex = 0.8,
     col = "black")

# Tambahkan legenda
legend("bottomleft",
       legend = c("r >= 0.8", "0.6 <= r < 0.8", "0.4 <= r < 0.6", "r < 0.4"),
       col = c("red", "orange", "gold", "gray70"),
       lwd = 3,
       bty = "n")
```

#uji heterogenitas

```r
# =========================================
# INDEKS GINI SESUAI RUMUS REFERENSI
# =========================================

gini_rumus <- function(K){

  # Hilangkan NA jika ada
  K <- K[!is.na(K)]

  # Urutkan dari terbesar ke terkecil
  K <- sort(K, decreasing = TRUE)

  # Banyak pengamatan
  n <- length(K)

  # Rata-rata
  K_bar <- mean(K)

  # Indeks urutan
  i <- 1:n

  # Menghitung Indeks Gini
  IG <- 1 + (1/n) -
    (2 / (n^2 * K_bar)) * sum(i * K)

  return(IG)
}
gini_K1 <- gini_rumus(kalimantanfixx$K1)
gini_K2 <- gini_rumus(kalimantanfixx$K2)
gini_K3 <- gini_rumus(kalimantanfixx$K3)
gini_K4 <- gini_rumus(kalimantanfixx$K4)
gini_K5 <- gini_rumus(kalimantanfixx$K5)

gini_K1
gini_K2
gini_K3
gini_K4
gini_K5
```
