#menyamakan total hari

```r
samakan_hari <- function(data){

  data$acq_date <- as.Date(data$acq_date)

  data_harian <- data %>%
    group_by(acq_date) %>%
    summarise(
      jumlah_hotspot = n()
    )

  hasil <- tanggal_lengkap %>%
    left_join(data_harian, by = "acq_date") %>%
    replace_na(list(jumlah_hotspot = 0))

  return(hasil)
}

kalbar_fix  <- samakan_hari(kalbar)
kalteng_fix <- samakan_hari(kalteng)
kalsel_fix  <- samakan_hari(kalsel)
kaltim_fix  <- samakan_hari(kaltim)
kaltara_fix <- samakan_hari(kaltara)
```

#uji heterogenitas

```r
# =========================================
# INDEKS GINI SESUAI RUMUS REFERENSI
# =========================================

gini_rumus <- function(K){

  # Hilangkan NA jika ada
  K <- K[!is.na(K)]

  # Urutkan dari terbesar ke terkecil
  K <- sort(K, decreasing = TRUE)

  # Banyak pengamatan
  n <- length(K)

  # Rata-rata
  K_bar <- mean(K)

  # Indeks urutan
  i <- 1:n

  # Menghitung Indeks Gini
  IG <- 1 + (1/n) -
    (2 / (n^2 * K_bar)) * sum(i * K)

  return(IG)
}
gini_K1 <- gini_rumus(kalimantanfixx$K1)
gini_K2 <- gini_rumus(kalimantanfixx$K2)
gini_K3 <- gini_rumus(kalimantanfixx$K3)
gini_K4 <- gini_rumus(kalimantanfixx$K4)
gini_K5 <- gini_rumus(kalimantanfixx$K5)

gini_K1
gini_K2
gini_K3
gini_K4
gini_K5
```

```r
library(forecast)
library(MASS)

# =========================================================
# 1. Ambil data & tambahkan konstanta agar semua nilai > 0
# =========================================================
data_k <- kalimantanfixx[, c("K1","K2","K3","K4","K5")]

sapply(data_k, min)      # nilai minimum awal (0)

konstanta <- 1
data_k <- data_k + konstanta

sapply(data_k, min)      # cek lagi, harus > 0 (yaitu 1)

# =========================================================
# 2. Estimasi lambda awal
# =========================================================
lambda_awal <- sapply(data_k, function(z) BoxCox.lambda(z, method = "loglik"))
lambda_awal

# =========================================================
# 3. Plot Box-Cox + selang kepercayaan 95% (opsional)
# =========================================================
par(mfrow = c(2, 3))
for (w in names(data_k)) {
  z <- data_k[[w]]
  boxcox(lm(z ~ 1), lambda = seq(-2, 2, 0.01),
         main = paste("Box-Cox", w))
}
par(mfrow = c(1, 1))

# =========================================================
# 4. Fungsi pembulatan & transformasi (sesuai Tabel 2.1)
# =========================================================
bulatkan_lambda <- function(l) {
  standar <- c(-1, -0.5, 0, 0.5, 1)
  standar[which.min(abs(standar - l))]
}

transformasi <- function(z, lam) {
  if (lam == -1)   return(1 / z)
  if (lam == -0.5) return(1 / sqrt(z))
  if (lam == 0)    return(log(z))
  if (lam == 0.5)  return(sqrt(z))
  if (lam == 1)    return(z)
}

nama_transf <- c("-1" = "1/z", "-0.5" = "1/sqrt(z)", "0" = "ln(z)",
                 "0.5" = "sqrt(z)", "1" = "z (tanpa transformasi)")

# =========================================================
# 5. Uji + transformasi untuk semua wilayah
# =========================================================
hasil <- data.frame()
data_transf <- data_k

for (w in names(data_k)) {
  z <- data_k[[w]]

  lam1  <- BoxCox.lambda(z, method = "loglik")
  lam1r <- bulatkan_lambda(lam1)

  z_t   <- transformasi(z, lam1r)

  # Setelah transformasi, nilai bisa <= 0 (misalnya ln(1) = 0),
  # jadi geser lagi agar positif sebelum cek lambda kedua
  z_cek <- if (min(z_t) <= 0) z_t - min(z_t) + 1 else z_t
  lam2  <- BoxCox.lambda(z_cek, method = "loglik")
  lam2r <- bulatkan_lambda(lam2)

  data_transf[[w]] <- z_t

  hasil <- rbind(hasil, data.frame(
    Wilayah            = w,
    Lambda_Awal        = round(lam1, 3),
    Lambda_Bulat_Awal  = lam1r,
    Transformasi       = nama_transf[as.character(lam1r)],
    Lambda_Setelah     = round(lam2, 3),
    Lambda_Bulat_Akhir = lam2r,
    Keterangan         = ifelse(lam2r == 1, "Stasioner", "Belum Stasioner")
  ))
}

rownames(hasil) <- NULL
print(hasil)

# =========================================================
# 6. Plot sebelum vs sesudah transformasi
# =========================================================
par(mfrow = c(2, 5))
for (w in names(data_k)) plot.ts(data_k[[w]], main = paste(w, "(+1)"), ylab = "")
for (w in names(data_k)) plot.ts(data_transf[[w]], main = paste(w, "(transf.)"), ylab = "")
par(mfrow = c(1, 1))
```

```r
library(forecast)
library(MASS)

# =========================================================
# 0. Data & fungsi dasar
# =========================================================
data_asli <- kalimantanfixx[, c("K1","K2","K3","K4","K5")]

bulatkan_lambda <- function(l) {
  standar <- c(-1, -0.5, 0, 0.5, 1)
  standar[which.min(abs(standar - l))]
}

transformasi <- function(z, lam) {
  if (lam == -1)   return(1 / z)
  if (lam == -0.5) return(1 / sqrt(z))
  if (lam == 0)    return(log(z))
  if (lam == 0.5)  return(sqrt(z))
  if (lam == 1)    return(z)
}

nama_transf <- c("-1" = "1/z", "-0.5" = "1/sqrt(z)", "0" = "ln(z)",
                 "0.5" = "sqrt(z)", "1" = "z (tanpa transformasi)")

# =========================================================
# 1. Fungsi transformasi ITERATIF sampai lambda = 1
# =========================================================
transform_iteratif <- function(z, konstanta = 1, maks_iter = 6) {
  z <- z + konstanta
  log_iter <- data.frame()
  stasioner <- FALSE

  for (i in 1:maks_iter) {
    lam  <- BoxCox.lambda(z, method = "loglik", lower = -2, upper = 2)
    lamr <- bulatkan_lambda(lam)

    log_iter <- rbind(log_iter, data.frame(
      Iterasi      = i,
      Lambda       = round(lam, 3),
      Lambda_Bulat = lamr,
      Transformasi = nama_transf[as.character(lamr)]
    ))

    if (lamr == 1) { stasioner <- TRUE; break }   # sudah stasioner
    if (i == maks_iter) break                      # batas iterasi

    z <- transformasi(z, lamr)
    if (min(z) <= 0) z <- z - min(z) + 1           # jaga agar tetap positif
  }
  list(data = z, log = log_iter, stasioner = stasioner)
}

# =========================================================
# 2. Jalankan untuk semua wilayah (konstanta = 1)
# =========================================================
hasil_iter <- lapply(data_asli, transform_iteratif, konstanta = 1)

# Lihat riwayat iterasi tiap wilayah
for (w in names(hasil_iter)) {
  cat("\n=== ", w, " ===\n")
  print(hasil_iter[[w]]$log, row.names = FALSE)
  cat("Status:", ifelse(hasil_iter[[w]]$stasioner, "Stasioner", "Belum stasioner"), "\n")
}

# Ringkasan
ringkasan <- data.frame(
  Wilayah    = names(hasil_iter),
  Iterasi    = sapply(hasil_iter, function(x) nrow(x$log)),
  Keterangan = sapply(hasil_iter, function(x) ifelse(x$stasioner, "Stasioner", "Belum Stasioner"))
)
print(ringkasan, row.names = FALSE)

# Data hasil transformasi
data_transf <- as.data.frame(lapply(hasil_iter, function(x) x$data))

# =========================================================
# 3. Jika masih ada yang belum stasioner: coba beberapa konstanta
# =========================================================
konstanta_coba <- c(0.1, 0.5, 1, 2, 5)

tabel_konstanta <- data.frame()
for (w in names(data_asli)) {
  for (k in konstanta_coba) {
    h <- transform_iteratif(data_asli[[w]], konstanta = k)
    tabel_konstanta <- rbind(tabel_konstanta, data.frame(
      Wilayah    = w,
      Konstanta  = k,
      Iterasi    = nrow(h$log),
      Lambda_Awal = h$log$Lambda[1],
      Keterangan = ifelse(h$stasioner, "Stasioner", "Belum")
    ))
  }
}
print(tabel_konstanta, row.names = FALSE)

# =========================================================
# 4. Cara lain memutuskan: selang kepercayaan 95% (MASS::boxcox)
#    Stasioner jika lambda = 1 ada di dalam selang kepercayaan
# =========================================================
cek_selang <- function(z, lam_uji = 1) {
  bc <- boxcox(lm(z ~ 1), lambda = seq(-3, 3, 0.01), plotit = FALSE)
  batas <- max(bc$y) - qchisq(0.95, 1) / 2
  ci <- range(bc$x[bc$y >= batas])
  c(Bawah = ci[1], Atas = ci[2],
    Lambda1_dalam_selang = lam_uji >= ci[1] & lam_uji <= ci[2])
}

selang <- t(sapply(data_transf, cek_selang))
print(round(selang, 3))

# =========================================================
# 5. Plot sebelum vs sesudah
# =========================================================
par(mfrow = c(2, 5))
for (w in names(data_asli)) plot.ts(data_asli[[w]], main = paste(w, "(asli)"), ylab = "")
for (w in names(data_transf)) plot.ts(data_transf[[w]], main = paste(w, "(transf.)"), ylab = "")
par(mfrow = c(1, 1))
```

```r
library(MASS)

cek_selang2 <- function(x, nama = "") {
  d  <- data.frame(y = as.numeric(x))
  bc <- boxcox(y ~ 1, data = d, lambda = seq(-3, 3, 0.01), plotit = FALSE)
  batas <- max(bc$y) - qchisq(0.95, 1) / 2
  ci <- range(bc$x[bc$y >= batas])
  data.frame(Wilayah = nama,
             Lambda_MLE = round(bc$x[which.max(bc$y)], 2),
             Bawah = round(ci[1], 2),
             Atas  = round(ci[2], 2),
             Lambda1_dalam_selang = 1 >= ci[1] & 1 <= ci[2])
}

# Cek satu kolom dulu
cek_selang2(data_transf$K1, "K1")

# Lalu semua kolom
do.call(rbind, lapply(names(data_transf),
        function(w) cek_selang2(data_transf[[w]], w)))
```
